```python
def get_nodes_by_ids(structure, node_ids):
    """
    Retrieve nodes by a list of node IDs.
    
    Args:
        structure: The tree structure (dict or list)
        node_ids: List of node IDs to retrieve (e.g., ['0001', '0005', '0010'])
    
    Returns:
        List of nodes matching the given IDs
    """
    all_nodes = structure_to_list(structure)
    node_id_set = set(node_ids)
    return [node for node in all_nodes if node.get('node_id') in node_id_set]


get_nodes_by_ids(result['structure'], ['0001', '0005', '0010'])
```
